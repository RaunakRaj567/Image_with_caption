```python
# caption processing

import re

def clean_caption(caption):

    # Convert everything to lowercase
    caption = caption.lower()

    # Remove punctuation and special characters
    caption = re.sub(r"[^a-z\s]", "", caption)

    # Remove extra spaces
    caption = re.sub(r"\s+", " ", caption).strip()

    # Capitalize only the first letter
    caption = caption.capitalize()

    return caption
```

```python
def pad_captions(sequences, max_length):

    padded = []

    for sequence in sequences:

        # Truncate if sequence is longer
        sequence = sequence[:max_length]

        # Add padding tokens
        sequence = sequence + [word_to_idx["<pad>"]] * (
            max_length - len(sequence)
        )

        padded.append(sequence)

    return torch.tensor(padded, dtype=torch.long)
```

```python
# custom dataset

from torch.utils.data import Dataset, DataLoader

class ImageCaptionDataset(Dataset):

    def __init__(self, dataframe, padded_captions, image_dir, transform):

        self.dataframe = dataframe.reset_index(drop=True)
        self.padded_captions = padded_captions
        self.image_dir = image_dir
        self.transform = transform

    def __len__(self):

        return len(self.dataframe)

    def __getitem__(self, index):

        # Get image filename
        image_name = self.dataframe.iloc[index]["image"]

        # Load image
        image_path = os.path.join(self.image_dir, image_name)

        img = image.open(image_path).convert("RGB")

        # Apply image transformations
        img = self.transform(img)

        # Get corresponding numerical caption
        caption = self.padded_captions[index]

        return img, caption
```

```python
# Build the CNN Encoder using ResNet50
import torch
import torch.nn as nn
from torchvision import models
```

```python
# Load pretrained ResNet50
weights = models.ResNet50_Weights.DEFAULT

resnet = models.resnet50(weights=weights)

print(resnet)
```

```python
# Remove final classification layer
resnet = nn.Sequential(*list(resnet.children())[:-1])

print(resnet)
```

```python
# Freeze all ResNet50 parameters
for param in resnet.parameters():
    param.requires_grad = False

print("ResNet50 parameters frozen successfully!")
```

```python
# creating feature extractor

class EncoderCNN(nn.Module):

    def __init__(self):
        super(EncoderCNN, self).__init__()

        self.resnet = resnet

    def forward(self, images):

        with torch.no_grad():
            features = self.resnet(images)

        # Flatten feature tensor
        features = features.view(features.size(0), -1)

        return features
```

```python
# Building LSTM Decoder

EMBED_SIZE = 256
HIDDEN_SIZE = 512
VOCAB_SIZE = len(word_to_idx)
NUM_LAYERS = 1

class DecoderLSTM(nn.Module):

    def __init__(self, embed_size, hidden_size, vocab_size):
        super(DecoderLSTM, self).__init__()

        # Convert word indexes into embeddings
        self.embedding = nn.Embedding(
            vocab_size,
            embed_size,
            padding_idx=word_to_idx["<pad>"]
        )

        # Convert CNN features into embedding size
        self.feature_projection = nn.Linear(2048, embed_size)

        # LSTM layer
        self.lstm = nn.LSTM(
            embed_size,
            hidden_size,
            NUM_LAYERS,
            batch_first=True
        )

        # Predict the next word
        self.fc = nn.Linear(hidden_size, vocab_size)

    def forward(self, features, captions):

        # Remove the last token from input captions
        embeddings = self.embedding(captions[:, :-1])

        # Convert image features into embeddings
        image_embeddings = self.feature_projection(features).unsqueeze(1)

        # Combine image information and caption embeddings
        inputs = torch.cat((image_embeddings, embeddings), dim=1)

        # Pass through LSTM
        outputs, _ = self.lstm(inputs)

        # Remove the image-only output
        outputs = outputs[:, 1:, :]

        # Generate vocabulary predictions
        outputs = self.fc(outputs)

        return outputs
```

```python
decoder = DecoderLSTM(
    EMBED_SIZE,
    HIDDEN_SIZE,
    VOCAB_SIZE
).to(device)

print("LSTM Decoder created successfully!")
```

```python
images, captions = next(iter(train_loader))

images = images.to(device)
captions = captions.to(device)

with torch.no_grad():
    features = encoder(images)
    outputs = decoder(features, captions)

print("Image features shape:", features.shape)
print("Input captions shape:", captions.shape)
print("Decoder output shape:", outputs.shape)
```

```python
# combining CN encode + LSTM decoder

class ImageCaptioningModel(nn.Module):

    def __init__(self, encoder, decoder):
        super(ImageCaptioningModel, self).__init__()

        self.encoder = encoder
        self.decoder = decoder

    def forward(self, images, captions):

        # Extract image features
        features = self.encoder(images)

        # Generate word predictions
        outputs = self.decoder(features, captions)

        return outputs
```

```python
model = ImageCaptioningModel(
    encoder,
    decoder
).to(device)

print("Complete Image Captioning Model created!")
```

```python
def train_one_epoch(model, train_loader, criterion, optimizer, device, scaler):

    model.train()

    # Keep pretrained CNN frozen
    model.encoder.eval()

    total_loss = 0

    for images, captions in tqdm(train_loader, desc="Training"):

        images = images.to(device)
        captions = captions.to(device)

        optimizer.zero_grad(set_to_none=True)

        # Extract image features using frozen CNN
        with torch.no_grad():
            with torch.autocast(
                device_type="cuda",
                dtype=torch.float16,
                enabled=(device.type == "cuda")
            ):
                features = model.encoder(images)

        # LSTM predictions
        with torch.autocast(
            device_type="cuda",
            dtype=torch.float16,
            enabled=(device.type == "cuda")
        ):
            predictions = model.decoder(features, captions)

            targets = captions[:, 1:]

            loss = criterion(
                predictions.reshape(-1, VOCAB_SIZE),
                targets.reshape(-1)
            )

        # Backpropagation
        scaler.scale(loss).backward()

        # Update decoder weights
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item()

    return total_loss / len(train_loader)
```

```python
def validate_one_epoch(model, val_loader, criterion, device):

    model.eval()

    total_loss = 0

    with torch.no_grad():

        for images, captions in tqdm(val_loader, desc="Validation"):

            images = images.to(device)
            captions = captions.to(device)

            with torch.autocast(
                device_type="cuda",
                dtype=torch.float16,
                enabled=(device.type == "cuda")
            ):
                predictions = model(images, captions)

                targets = captions[:, 1:]

                loss = criterion(
                    predictions.reshape(-1, VOCAB_SIZE),
                    targets.reshape(-1)
                )

            total_loss += loss.item()

    return total_loss / len(val_loader)
```

```python
from tqdm.notebook import tqdm

EPOCHS = 15
PATIENCE = 3

train_losses = []
val_losses = []

best_val_loss = float("inf")
patience_counter = 0

os.makedirs(PROJECT_DIR, exist_ok=True)

CHECKPOINT_PATH = os.path.join(
    PROJECT_DIR,
    "best_image_caption_model.pth"
)

for epoch in range(EPOCHS):

    print(f"\n{'='*45}")
    print(f"Epoch [{epoch + 1}/{EPOCHS}]")
    print(f"{'='*45}")

    # Training
    train_loss = train_one_epoch(
        model,
        train_loader,
        criterion,
        optimizer,
        device,
        scaler
    )

    # Validation
    val_loss = validate_one_epoch(
        model,
        val_loader,
        criterion,
        device
    )

    train_losses.append(train_loss)
    val_losses.append(val_loss)

    print(f"Training Loss:   {train_loss:.4f}")
    print(f"Validation Loss: {val_loss:.4f}")

    # Save best model
    if val_loss < best_val_loss:

        best_val_loss = val_loss
        patience_counter = 0

        torch.save({
            "epoch": epoch + 1,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "best_val_loss": best_val_loss
        }, CHECKPOINT_PATH)

        print("Best model checkpoint saved!")

    else:
        patience_counter += 1
        print(f"No improvement: {patience_counter}/{PATIENCE}")

    # Early stopping
    if patience_counter >= PATIENCE:
        print("\nEarly stopping triggered!")
        break

print("\nTraining completed!")
print("Best validation loss:", best_val_loss)
print("Checkpoint:", CHECKPOINT_PATH)
```

```python
def generate_caption(img_tensor, model, method="beam", beam_width=5, max_length=MAX_LENGTH):
    """
    Generates a caption for an input image tensor using either:
    - 'beam': Beam Search with length penalty and repetition reduction (High Quality)
    - 'greedy': Aligned Greedy Search (Fast)
    """
    model.eval()

    if img_tensor.dim() == 3:
        img_tensor = img_tensor.unsqueeze(0)
    img_tensor = img_tensor.to(device)

    start_token = word_to_idx["<start>"]
    end_token = word_to_idx["<end>"]

    with torch.no_grad():
        features = model.encoder(img_tensor)
        img_emb = model.decoder.feature_projection(features).unsqueeze(1)
        _, states = model.decoder.lstm(img_emb)

        if method == "greedy":
            generated_words = []
            curr_token = torch.tensor([[start_token]], dtype=torch.long, device=device)

            for _ in range(max_length):
                token_emb = model.decoder.embedding(curr_token)
                output, states = model.decoder.lstm(token_emb, states)
                logits = model.decoder.fc(output.squeeze(1))
                predicted_id = logits.argmax(dim=1).item()

                if predicted_id == end_token:
                    break

                word = idx_to_word[predicted_id]
                if word not in ["<start>", "<pad>", "<unk>"]:
                    generated_words.append(word)

                curr_token = torch.tensor([[predicted_id]], dtype=torch.long, device=device)

            return " ".join(generated_words).capitalize()

        else:
            # Beam Search with width k=beam_width
            sequences = [(0.0, [start_token], states)]

            for _ in range(max_length):
                all_candidates = []
                all_completed = True

                for score, seq, st in sequences:
                    if seq[-1] == end_token:
                        all_candidates.append((score, seq, st))
                        continue

                    all_completed = False
                    last_token = torch.tensor([[seq[-1]]], dtype=torch.long, device=device)
                    emb = model.decoder.embedding(last_token)
                    output, new_st = model.decoder.lstm(emb, st)
                    logits = model.decoder.fc(output.squeeze(1))
                    log_probs = torch.log_softmax(logits, dim=1).squeeze(0)

                    # Repetition penalty
                    for past_tok in set(seq[1:]):
                        log_probs[past_tok] -= 1.2

                    topk_probs, topk_indices = torch.topk(log_probs, beam_width)
                    for k in range(beam_width):
                        cand_score = score + topk_probs[k].item()
                        cand_seq = seq + [topk_indices[k].item()]
                        all_candidates.append((cand_score, cand_seq, new_st))

                if all_completed:
                    break

                # Length normalized ranking
                all_candidates.sort(key=lambda x: x[0] / (len(x[1]) ** 0.7), reverse=True)
                sequences = all_candidates[:beam_width]

            best_seq = sequences[0][1]
            words = [
                idx_to_word[tok]
                for tok in best_seq
                if tok not in [start_token, end_token, word_to_idx["<pad>"], word_to_idx["<unk>"]]
            ]
            return " ".join(words).capitalize()

```
